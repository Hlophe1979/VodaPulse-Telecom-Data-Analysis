# VodaPulse Synthetic Dataset Generation

```python
# ==========================================
# GENERATE DATASET FILE FIRST
# ==========================================
import random
from datetime import datetime, timedelta
import numpy as np
import pandas as pd

np.random.seed(42)
random.seed(42)

n_records = 5000

start_date = datetime(2026, 1, 1)
date_strings = [
    (start_date + timedelta(days=random.randint(0, 180))).strftime("%Y-%m-%d")
    for _ in range(n_records)
]

subscriber_ids = [f"VP_{10000 + i}" for i in range(n_records)]
customer_age = np.random.randint(18, 75, size=n_records)

provinces = np.random.choice(
    ["Gauteng", "KwaZulu-Natal", "Western Cape", "Eastern Cape", "Limpopo"],
    n_records,
    p=[0.35, 0.22, 0.18, 0.13, 0.12],
)

area_types = np.random.choice(
    ["Urban", "Suburban", "Rural"], n_records, p=[0.55, 0.30, 0.15]
)

contract_types = np.random.choice(
    ["Prepaid", "Postpaid Contract", "Enterprise Hybrid"],
    n_records,
    p=[0.60, 0.30, 0.10],
)

device_types = np.random.choice(
    ["5G Smartphone", "4G Smartphone", "LTE Router/MiFi", "Legacy Feature Phone"],
    n_records,
    p=[0.40, 0.45, 0.10, 0.05],
)

payment_methods = np.random.choice(
    ["Debit Order", "Voucher Top-Up", "Mobile App / EFT", "Credit Card"],
    n_records,
    p=[0.35, 0.40, 0.15, 0.10],
)

network_experience = np.random.choice(
    ["Excellent", "Good", "Fair", "Poor / Frequent Drops"],
    n_records,
    p=[0.40, 0.35, 0.15, 0.10],
)

support_tickets = np.random.choice(
    ["None", "Billing Query", "Network Coverage", "SIM Swap / Technical"],
    n_records,
    p=[0.65, 0.15, 0.12, 0.08],
)

load_shedding_impact = np.random.choice(
    ["Severe (Stage 4+)", "Moderate (Stage 1-3)", "Minimal / No Impact"],
    n_records,
    p=[0.30, 0.50, 0.20],
)

app_engagement_level = np.random.choice(
    ["High Active", "Moderate", "Low / Inactive"],
    n_records,
    p=[0.30, 0.45, 0.25],
)

monthly_usage_gb = np.round(np.random.gamma(shape=2, scale=10, size=n_records), 1)
arpu_zar = np.round(
    monthly_usage_gb * np.random.uniform(12, 22) + np.random.uniform(40, 150), 2
)

df = pd.DataFrame(
    {
        "Date": date_strings,
        "Subscriber_ID": subscriber_ids,
        "Customer_Age": customer_age,
        "Province": provinces,
        "Area_Type": area_types,
        "Contract_Type": contract_types,
        "Device_Type": device_types,
        "Payment_Method": payment_methods,
        "Network_Experience": network_experience,
        "Support_Ticket_Type": support_tickets,
        "Load_Shedding_Impact": load_shedding_impact,
        "App_Engagement": app_engagement_level,
        "Data_Usage_GB": monthly_usage_gb,
        "ARPU_ZAR": arpu_zar,
    }
)

df.to_csv("vodapulse_raw_dataset.csv", index=False)
print("File 'vodapulse_raw_dataset.csv' created successfully!")
```
